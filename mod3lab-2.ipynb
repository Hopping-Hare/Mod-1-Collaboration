```python
# BellaJane Brimm
# mod3lab.ipynb
# This program utilizes classes and functions to make a mini app that takes 
# and outputs the information of their vehicle.

class Vehicle:
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type

class Automobile(Vehicle):
    def __init__(self, year, make, model, doors, roof):
        super().__init__("car")
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof

year = int(input("What year is your car?: "))
make = input("What make is your car?: ")
model = input("What model is your car?: ")
doors = int(input("How many doors are there? (2 or 4): "))
roof = input("What type of roof does it have? (solid or sun): ")

car = Automobile(year, make, model, doors, roof)

print("Vehicle Type:", car.vehicle_type)
print("Year:", car.year)
print("Make:", car.make)
print("Model:", car.model)
print("Number of doors:", car.doors)
print("Type of roof:", car.roof)
```
