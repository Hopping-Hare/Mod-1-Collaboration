```python
# BellaJane
# Student Records
# This program takes your first and last name, as well as your gpa,
# and determines if you made the Dean's L


first_name = input("What is your first name?")
last_name = input("What is your last name?")
gpa = float(input("What is your GPA?"))

while last_name != "ZZZ":
    if gpa >= 3.5:
        print("You have made the Dean's List.")
        break
    if gpa >= 3.25:
        print("You have made the Honor Roll.")
        break
    if gpa < 3.25:
        print("Sorry, but you did not make either one.")
        break
```
